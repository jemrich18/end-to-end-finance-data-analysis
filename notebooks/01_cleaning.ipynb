# Phase 1: Data Cleaning
In this notebook we load, inspect, and clean the Personal Finance dataset 
in preparation for SQL analysis and visualization.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/Personal_Finance_Dataset.csv')
df.head()
```

```python
# Check shape, column names, and data types
print("Shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
print("\nData types:\n", df.dtypes)
```

```python
# Identify any missing data
print("Missing values per column:")
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Unique categories:\n", df['Category'].unique())
print("\nTransaction types:\n", df['Type'].unique())
```

## Observations
- No missing values or duplicate rows found
- Date column is stored as string type — needs converting to datetime
- Categories and Type values are clean and consistent
- Amount is float64 which is correct for currency values

```python
# Convert date from string to datetime 
df['Date'] = pd.to_datetime(df['Date'])

print("Date column type after conversion:", df['Date'].dtype)
print("Date range:", df['Date'].min(), "to", df['Date'].max())
```

```python
# Extract month, year and day of week for analysis later
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Month_Name'] = df['Date'].dt.strftime('%B')
df['Day_of_Week'] = df['Date'].dt.strftime('%A')

df.head()
```

```python
# Remove spaces from column names — makes SQL and pandas easier to work with
df.columns = df.columns.str.replace(' ', '_')

print("Updated column names:", df.columns.tolist())
```

```python
print("Final shape:", df.shape)
print("\nFinal dtypes:\n", df.dtypes)
df.head()
```

```python
# Save cleaned dataset for use in next phases
df.to_csv('../data/cleaned_finance_data.csv', index=False)
print("Cleaned data saved successfully!")
```

## Summary
- Converted Date column from string to datetime
- Extracted Year, Month, Month_Name and Day_of_Week columns for time-based analysis
- Renamed columns to remove spaces for easier querying
- Saved cleaned dataset to `data/cleaned_finance_data.csv`

**No major data quality issues found. Dataset is ready for analysis.**
